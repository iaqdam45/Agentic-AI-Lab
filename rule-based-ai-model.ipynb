```python
def weather_advice(temp_c, humidity_percent, wind_speed, rain_chance):
    advice = []

    if temp_c >= 40:
        advice.append("Extreme heat: stay indoors")
    elif temp_c >= 35:
        advice.append("Very hot: stay hydrated")
    elif temp_c >= 30:
        advice.append("Hot weather: drink plenty of water")
    elif temp_c >= 20:
        advice.append("Temperature is comfortable")
    elif temp_c >= 10:
        advice.append("Cool weather: wear a light jacket")
    elif temp_c >= 0:
        advice.append("Cold weather: wear a warm jacket")
    else:
        advice.append("Freezing weather: stay warm")

    if humidity_percent >= 90:
        advice.append("Extremely high humidity: it may feel very uncomfortable")
    elif humidity_percent >= 80:
        advice.append("High humidity: expect rain")
    elif humidity_percent >= 60:
        advice.append("Moderate humidity")
    elif humidity_percent <= 20:
        advice.append("Very low humidity")
    else:
        advice.append("Humidity is at a comfortable level")

    # Wind speed rules
    if wind_speed >= 50:
        advice.append("Very strong winds: avoid outdoor activities")
    elif wind_speed >= 30:
        advice.append("Strong winds: be careful outdoors")
    else:
        advice.append("Wind speed is normal")

    # Rain chance rules
    if rain_chance >= 80:
        advice.append("Very high chance of rain: carry an umbrella")
    elif rain_chance >= 50:
        advice.append("Moderate chance of rain: consider carrying an umbrella")
    else:
        advice.append("Low chance of rain")

    return advice


if __name__ == "__main__":
    sample_days = [(32, 58, 14, 25), (25, 62, 9, 35), (18, 48, 16, 15), (9, 67, 11, 45), (3, 54, 7, 20)]

    for temp, humidity, wind, rain in sample_days:
        result = weather_advice(temp, humidity, wind, rain)
        print(f"Temp = {temp}C, Humidity = {humidity}%, Wind = {wind} km/h, Rain chance = {rain}% -> {result}")
```

```python
def predict_next_day(past_temp, window=3):
    recent = past_temp[-window:]
    prediction = sum(recent) / len(recent)
    return round(prediction, 1)

if __name__ == "__main__":
    past_week = [30,31,29,32,33,31,30]
    actual_tomorrow = 31
    predicted = predict_next_day(past_week, window=3)
    error = round(abs(predicted - actual_tomorrow), 1)
    print(f"Predicted: {predicted}C")
    print(f"Actual: {actual_tomorrow}C")
    print(f"Error:{error}C")
    
```

```python
def classify_email(email, keywords, threshold=2):
    count = 0
    for keyword in keywords:
        if keyword in email.lower():
            count += 1

    if count >= threshold:
        return "Spam"
    else:
        return "Legitimate"

# Classification assigns an input to one of a fixed set of categories.
spam_keywords = ["free", "winner", "urgent", "click here", "prize"]
emails = [
    ("Congratulations! You are a winner and have won a prize!", "Spam"),
    ("Urgent! Click here to claim your free prize!", "Spam"),
    ("You are the winner of our free shopping prize.", "Spam"),
    ("Hi, can we meet tomorrow to discuss the project?", "Legitimate"),
    ("Your monthly electricity bill is ready to view.", "Legitimate"),
    ("Thank you for submitting your assignment. We will review it soon.", "Legitimate")
]
correct = 0

for email, actual_label in emails:
    predicted_label = classify_email(email, spam_keywords)

    print("Email:", email)
    print("Predicted:", predicted_label)
    print("Actual:", actual_label)
    print()

    if predicted_label == actual_label:
        correct += 1

accuracy = (correct / len(emails)) * 100

print("Correct classifications:", correct, "out of", len(emails))
print("Accuracy:", accuracy, "%")
```

```python
def classify_email(email, keywords, threshold):
    count = 0
    for keyword in keywords:
        if keyword in email.lower():
            count += 1
    if count >= threshold:
        return "Spam"
    else:
        return "Legitimate"
spam_keywords = ["free", "winner", "urgent", "click here", "prize"]
emails = [
    ("Congratulations! You are a winner and have won a prize!", "Spam"),
    ("Urgent! Click here to claim your free prize!", "Spam"),
    ("You are the winner of our free shopping prize.", "Spam"),
    ("Hi, can we meet tomorrow to discuss the project?", "Legitimate"),
    ("Your monthly electricity bill is ready to view.", "Legitimate"),
    ("Thank you for submitting your assignment. We will review it soon.", "Legitimate")
]
for threshold in [1, 2, 3, 4]:
    correct = 0
    for email, actual_label in emails:
        predicted_label = classify_email(
            email, spam_keywords, threshold
        )
        if predicted_label == actual_label:
            correct += 1
    accuracy = (correct / len(emails)) * 100
    print("Threshold:", threshold)
    print("Accuracy:", accuracy, "%")
    print()
```
